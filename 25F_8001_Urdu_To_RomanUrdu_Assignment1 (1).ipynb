```python
import os
import re
import pandas as pd
from tqdm import tqdm
from sklearn.model_selection import train_test_split


BASE_DIR = 'urdu_ghazals_rekhta/data/dataset'
OUT_DIR = 'urdu_roman_output'
os.makedirs(OUT_DIR, exist_ok=True)


URDU_ROMAN_MAP = {
    'ا': 'a', 'آ': 'aa', 'ب': 'b', 'پ': 'p', 'ت': 't', 'ٹ': 't', 'ث': 's',
    'ج': 'j', 'چ': 'ch', 'ح': 'h', 'خ': 'kh', 'د': 'd', 'ڈ': 'd', 'ذ': 'z',
    'ر': 'r', 'ڑ': 'r', 'ز': 'z', 'ژ': 'zh', 'س': 's', 'ش': 'sh', 'ص': 's',
    'ض': 'z', 'ط': 't', 'ظ': 'z', 'ع': '', 'غ': 'gh', 'ف': 'f', 'ق': 'q',
    'ک': 'k', 'گ': 'g', 'ل': 'l', 'م': 'm', 'ن': 'n', 'ں': 'n', 'و': 'o',
    'ہ': 'h', 'ھ': 'h', 'ء': '', 'ی': 'i', 'ے': 'e',
    '،': '', '۔': '', '؟': '', '!': ''
}

def urdu_to_roman(text):
    result = []
    for ch in text:
        result.append(URDU_ROMAN_MAP.get(ch, ch))
    txt = ''.join(result)


    txt = re.sub(r'\s+', ' ', txt)
    txt = txt.replace('aa', 'a').replace('ee', 'i')
    txt = txt.strip()
    return txt



def load_urdu_lines(base):
    lines = []

    poets = [f for f in os.listdir(base)
             if os.path.isdir(os.path.join(base, f)) and not f.startswith('.')]

    for poet in tqdm(poets, desc="Loading Urdu"):
        ur_dir = os.path.join(base, poet, 'ur')

        if not os.path.exists(ur_dir):
            continue

        files = [f for f in os.listdir(ur_dir)
                 if os.path.isfile(os.path.join(ur_dir, f))]

        for fname in files:
            path = os.path.join(ur_dir, fname)

            try:
                with open(path, 'r', encoding='utf-8') as f:
                    for line in f:
                        line = re.sub(r'^\d+\s+', '', line.strip())
                        if len(line) > 3:
                            lines.append({
                                'urdu': line,
                                'poet': poet,
                                'file': fname
                            })
            except:
                continue

    return lines


print(" Reading Urdu texts...")
records = load_urdu_lines(BASE_DIR)
print(f" Loaded {len(records)} Urdu lines")

if not records:
    raise SystemExit(" No Urdu data found")

df = pd.DataFrame(records)

print(" Generating Roman Urdu...")
df['roman'] = df['urdu'].apply(urdu_to_roman)

full_csv = os.path.join(OUT_DIR, 'urdu_roman_full.csv')
df.to_csv(full_csv, index=False, encoding='utf-8-sig')
print(f" Full CSV saved: {full_csv}")

train_df, temp_df = train_test_split(df, test_size=0.2, random_state=42)
val_df, test_df = train_test_split(temp_df, test_size=0.5, random_state=42)

train_df.to_csv(os.path.join(OUT_DIR, 'train.csv'), index=False, encoding='utf-8-sig')
val_df.to_csv(os.path.join(OUT_DIR, 'val.csv'), index=False, encoding='utf-8-sig')
test_df.to_csv(os.path.join(OUT_DIR, 'test.csv'), index=False, encoding='utf-8-sig')

print("\n Dataset split complete:")
print(f"Train: {len(train_df)} rows")
print(f"Validation: {len(val_df)} rows")
print(f"Test: {len(test_df)} rows")

# Show sample
print("\n Sample rows:")
print(df[['urdu', 'roman']].head())
```

```python
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader
import numpy as np
import pandas as pd
import re
import json
import pickle
from collections import Counter, defaultdict
import math
import random
from typing import List, Tuple, Dict
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from tqdm import tqdm
import os
import time


torch.manual_seed(42)
np.random.seed(42)
random.seed(42)

class BPETokenizer:
    """
    Custom Byte-Pair Encoding (BPE) Tokenizer implemented from scratch
    """

    def __init__(self, vocab_size=2000):
        self.vocab_size = vocab_size
        self.vocab = {}
        self.merges = []
        self.idx_to_token = {}
        self.trained = False

    def get_word_tokens(self, text):
        """Split text into words and convert to character-level tokens"""
        words = text.strip().split()
        word_tokens = {}

        for word in words:

            chars = list(word) + ['</w>']
            token_word = ' '.join(chars)
            word_tokens[token_word] = word_tokens.get(token_word, 0) + 1

        return word_tokens

    def get_pairs(self, word_tokens):
        """Get all pairs of adjacent tokens with their frequencies"""
        pairs = defaultdict(int)

        for word, freq in word_tokens.items():
            symbols = word.split()
            for i in range(len(symbols) - 1):
                pair = (symbols[i], symbols[i + 1])
                pairs[pair] += freq

        return pairs

    def merge_tokens(self, pair, word_tokens):
        """Merge the most frequent pair in all words"""
        new_word_tokens = {}
        bigram = ' '.join(pair)
        replacement = ''.join(pair)

        for word in word_tokens:
            new_word = word.replace(bigram, replacement)
            new_word_tokens[new_word] = word_tokens[word]

        return new_word_tokens

    def train(self, corpus):
        """Train BPE on the given corpus"""
        print(f" Training BPE tokenizer (vocab_size: {self.vocab_size})...")

        word_tokens = self.get_word_tokens(corpus)

        vocab = set()
        for word in word_tokens:
            for token in word.split():
                vocab.add(token)

        print(f"  Initial vocab size: {len(vocab)}")

        num_merges = self.vocab_size - len(vocab) - 4

        for i in tqdm(range(num_merges), desc="  BPE Merges"):
            pairs = self.get_pairs(word_tokens)
            if not pairs:
                break
            best_pair = max(pairs.items(), key=lambda x: x[1])[0]

            word_tokens = self.merge_tokens(best_pair, word_tokens)
            self.merges.append(best_pair)
            vocab.add(''.join(best_pair))

        sorted_vocab = sorted(list(vocab))
        self.vocab = {token: i for i, token in enumerate(sorted_vocab)}


        self.vocab['<PAD>'] = len(self.vocab)
        self.vocab['<SOS>'] = len(self.vocab)
        self.vocab['<EOS>'] = len(self.vocab)
        self.vocab['<UNK>'] = len(self.vocab)


        self.idx_to_token = {i: token for token, i in self.vocab.items()}

        self.trained = True
        print(f"   BPE training complete! Final vocab size: {len(self.vocab)}")

    def encode(self, text):
        """Encode text using trained BPE"""
        if not self.trained:
            raise ValueError("Tokenizer not trained yet!")


        word_tokens = self.get_word_tokens(text)


        for pair in self.merges:
            word_tokens = self.merge_tokens(pair, word_tokens)

        tokens = []
        for word in word_tokens:
            for token in word.split():
                tokens.append(self.vocab.get(token, self.vocab['<UNK>']))

        return tokens

    def decode(self, indices):
        """Decode indices back to text"""
        tokens = []
        for idx in indices:
            if idx in self.idx_to_token and idx not in [
                self.vocab['<PAD>'], self.vocab['<SOS>'],
                self.vocab['<EOS>'], self.vocab['<UNK>']
            ]:
                tokens.append(self.idx_to_token[idx])


        text = ''.join(tokens)
        text = text.replace('</w>', ' ')
        return text.strip()

class UrduRomanDataset(Dataset):
    """
    Dataset class for Urdu to Roman Urdu translation pairs
    """

    def __init__(self, urdu_texts, roman_texts, urdu_tokenizer, roman_tokenizer, max_length=50):
        self.urdu_texts = urdu_texts
        self.roman_texts = roman_texts
        self.urdu_tokenizer = urdu_tokenizer
        self.roman_tokenizer = roman_tokenizer
        self.max_length = max_length

    def __len__(self):
        return len(self.urdu_texts)

    def __getitem__(self, idx):
        urdu_text = str(self.urdu_texts[idx])
        roman_text = str(self.roman_texts[idx])


        urdu_tokens = self.urdu_tokenizer.encode(urdu_text)
        roman_tokens = self.roman_tokenizer.encode(roman_text)

        urdu_tokens = urdu_tokens[:self.max_length-2]
        roman_tokens = roman_tokens[:self.max_length-2]


        urdu_tokens = ([self.urdu_tokenizer.vocab['<SOS>']] +
                      urdu_tokens +
                      [self.urdu_tokenizer.vocab['<EOS>']])

        roman_tokens = ([self.roman_tokenizer.vocab['<SOS>']] +
                       roman_tokens +
                       [self.roman_tokenizer.vocab['<EOS>']])


        while len(urdu_tokens) < self.max_length:
            urdu_tokens.append(self.urdu_tokenizer.vocab['<PAD>'])

        while len(roman_tokens) < self.max_length:
            roman_tokens.append(self.roman_tokenizer.vocab['<PAD>'])

        return torch.tensor(urdu_tokens), torch.tensor(roman_tokens)

class BiLSTMEncoder(nn.Module):
    """
    Bidirectional LSTM Encoder as specified in assignment
    """

    def __init__(self, vocab_size, embed_dim, hidden_dim, num_layers, dropout):
        super(BiLSTMEncoder, self).__init__()

        self.embedding = nn.Embedding(vocab_size, embed_dim, padding_idx=0)
        self.lstm = nn.LSTM(
            embed_dim, hidden_dim, num_layers,
            batch_first=True, bidirectional=True,
            dropout=dropout if num_layers > 1 else 0
        )
        self.dropout = nn.Dropout(dropout)

    def forward(self, x):

        embedded = self.dropout(self.embedding(x))


        outputs, (hidden, cell) = self.lstm(embedded)


        return outputs, (hidden, cell)

class LSTMDecoder(nn.Module):
    """
    LSTM Decoder as specified in assignment
    """

    def __init__(self, vocab_size, embed_dim, hidden_dim, num_layers, dropout, encoder_hidden_dim):
        super(LSTMDecoder, self).__init__()

        self.vocab_size = vocab_size
        self.hidden_dim = hidden_dim
        self.num_layers = num_layers

        self.embedding = nn.Embedding(vocab_size, embed_dim, padding_idx=0)
        self.lstm = nn.LSTM(
            embed_dim, hidden_dim, num_layers,
            batch_first=True,
            dropout=dropout if num_layers > 1 else 0
        )
        self.out = nn.Linear(hidden_dim, vocab_size)
        self.dropout = nn.Dropout(dropout)


        self.hidden_projection = nn.Linear(encoder_hidden_dim * 2, hidden_dim)
        self.cell_projection = nn.Linear(encoder_hidden_dim * 2, hidden_dim)

    def forward(self, x, hidden, cell):

        embedded = self.dropout(self.embedding(x))
        output, (hidden, cell) = self.lstm(embedded, (hidden, cell))

        output = self.out(output)


        return output, (hidden, cell)

    def init_hidden(self, encoder_hidden, encoder_cell):

        batch_size = encoder_hidden.size(1)
        encoder_layers = encoder_hidden.size(0) // 2

        encoder_hidden = encoder_hidden.view(encoder_layers, 2, batch_size, -1)
        encoder_cell = encoder_cell.view(encoder_layers, 2, batch_size, -1)

        last_hidden = torch.cat([encoder_hidden[-1, 0], encoder_hidden[-1, 1]], dim=1)
        last_cell = torch.cat([encoder_cell[-1, 0], encoder_cell[-1, 1]], dim=1)


        hidden = self.hidden_projection(last_hidden).unsqueeze(0).repeat(self.num_layers, 1, 1)
        cell = self.cell_projection(last_cell).unsqueeze(0).repeat(self.num_layers, 1, 1)

        return hidden.contiguous(), cell.contiguous()

class Seq2SeqModel(nn.Module):


    def __init__(self, encoder_vocab_size, decoder_vocab_size, embed_dim,
                 hidden_dim, encoder_layers, decoder_layers, dropout):
        super(Seq2SeqModel, self).__init__()

        self.encoder = BiLSTMEncoder(
            encoder_vocab_size, embed_dim, hidden_dim, encoder_layers, dropout
        )
        self.decoder = LSTMDecoder(
            decoder_vocab_size, embed_dim, hidden_dim, decoder_layers, dropout, hidden_dim
        )

    def forward(self, src, tgt, teacher_forcing_ratio=0.5):
        batch_size = src.size(0)
        tgt_len = tgt.size(1)
        tgt_vocab_size = self.decoder.vocab_size

        outputs = torch.zeros(batch_size, tgt_len, tgt_vocab_size).to(src.device)

        encoder_outputs, (hidden, cell) = self.encoder(src)

        hidden, cell = self.decoder.init_hidden(hidden, cell)

        input_token = tgt[:, 0].unsqueeze(1)

        for t in range(1, tgt_len):

            output, (hidden, cell) = self.decoder(input_token, hidden, cell)
            outputs[:, t] = output.squeeze(1)

            use_teacher_forcing = random.random() < teacher_forcing_ratio

            top1 = output.argmax(dim=2)
            input_token = tgt[:, t].unsqueeze(1) if use_teacher_forcing else top1

        return outputs
```

```python

def calculate_bleu(reference, candidate, n=4):

    from collections import Counter
    import math

    def get_ngrams(tokens, n):
        return [tuple(tokens[i:i+n]) for i in range(len(tokens) - n + 1)]

    ref_tokens = reference.split()
    cand_tokens = candidate.split()

    if len(cand_tokens) == 0:
        return 0.0

    precisions = []
    for i in range(1, min(n, len(cand_tokens)) + 1):
        ref_ngrams = Counter(get_ngrams(ref_tokens, i))
        cand_ngrams = Counter(get_ngrams(cand_tokens, i))

        matches = sum(min(ref_ngrams[gram], cand_ngrams[gram]) for gram in cand_ngrams)
        total = sum(cand_ngrams.values())

        if total == 0:
            precisions.append(0.0)
        else:
            precisions.append(matches / total)

    if min(precisions) == 0:
        return 0.0

    bp = min(1.0, len(cand_tokens) / len(ref_tokens)) if len(ref_tokens) > 0 else 0

    log_precisions = [math.log(p) for p in precisions if p > 0]
    if len(log_precisions) == 0:
        return 0.0

    return bp * math.exp(sum(log_precisions) / len(log_precisions))

def calculate_cer(reference, hypothesis):

    def levenshtein_distance(s1, s2):
        if len(s1) < len(s2):
            return levenshtein_distance(s2, s1)

        if len(s2) == 0:
            return len(s1)

        previous_row = list(range(len(s2) + 1))
        for i, c1 in enumerate(s1):
            current_row = [i + 1]
            for j, c2 in enumerate(s2):
                insertions = previous_row[j + 1] + 1
                deletions = current_row[j] + 1
                substitutions = previous_row[j] + (c1 != c2)
                current_row.append(min(insertions, deletions, substitutions))
            previous_row = current_row

        return previous_row[-1]

    if len(reference) == 0:
        return 1.0 if len(hypothesis) > 0 else 0.0

    distance = levenshtein_distance(reference, hypothesis)
    return distance / len(reference)

def calculate_perplexity(model, data_loader, criterion, device):
    """Calculate perplexity"""
    model.eval()
    total_loss = 0
    total_tokens = 0

    with torch.no_grad():
        for src, tgt in data_loader:
            src, tgt = src.to(device), tgt.to(device)

            output = model(src, tgt, teacher_forcing_ratio=0)

            output = output[:, 1:].contiguous().view(-1, output.size(-1))
            tgt = tgt[:, 1:].contiguous().view(-1)

            mask = (tgt != 0)
            if mask.sum() > 0:
                loss = criterion(output[mask], tgt[mask])
                total_loss += loss.item() * mask.sum().item()
                total_tokens += mask.sum().item()

    if total_tokens == 0:
        return float('inf')

    avg_loss = total_loss / total_tokens
    perplexity = math.exp(avg_loss)
    return perplexity

def train_model(model, train_loader, val_loader, num_epochs, learning_rate, device, pad_token_id):
    """Train the seq2seq model"""
    print(f" Training model for {num_epochs} epochs...")

    criterion = nn.CrossEntropyLoss(ignore_index=pad_token_id)
    optimizer = optim.Adam(model.parameters(), lr=learning_rate)
    scheduler = optim.lr_scheduler.ReduceLROnPlateau(optimizer, mode='min', factor=0.8, patience=2)

    train_losses = []
    val_losses = []
    val_perplexities = []

    best_val_loss = float('inf')
    patience = 5
    patience_counter = 0

    for epoch in range(num_epochs):

        model.train()
        epoch_train_loss = 0
        num_batches = 0

        progress_bar = tqdm(train_loader, desc=f"Epoch {epoch+1}/{num_epochs}")

        for batch_idx, (src, tgt) in enumerate(progress_bar):
            src, tgt = src.to(device), tgt.to(device)

            optimizer.zero_grad()

            output = model(src, tgt, teacher_forcing_ratio=0.7)

            output = output[:, 1:].contiguous().view(-1, output.size(-1))
            tgt = tgt[:, 1:].contiguous().view(-1)

            loss = criterion(output, tgt)
            loss.backward()

            torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)

            optimizer.step()

            epoch_train_loss += loss.item()
            num_batches += 1

            progress_bar.set_postfix({
                'Loss': f'{loss.item():.4f}',
                'Avg Loss': f'{epoch_train_loss/num_batches:.4f}'
            })

        avg_train_loss = epoch_train_loss / num_batches

        model.eval()
        epoch_val_loss = 0
        val_batches = 0

        with torch.no_grad():
            for src, tgt in val_loader:
                src, tgt = src.to(device), tgt.to(device)

                output = model(src, tgt, teacher_forcing_ratio=0)

                output = output[:, 1:].contiguous().view(-1, output.size(-1))
                tgt = tgt[:, 1:].contiguous().view(-1)

                loss = criterion(output, tgt)
                epoch_val_loss += loss.item()
                val_batches += 1

        avg_val_loss = epoch_val_loss / val_batches if val_batches > 0 else float('inf')

        val_perplexity = calculate_perplexity(model, val_loader, criterion, device)

        train_losses.append(avg_train_loss)
        val_losses.append(avg_val_loss)
        val_perplexities.append(val_perplexity)


        scheduler.step(avg_val_loss)

        print(f"Epoch {epoch+1}: Train Loss = {avg_train_loss:.4f}, "
              f"Val Loss = {avg_val_loss:.4f}, Val Perplexity = {val_perplexity:.2f}")

        if avg_val_loss < best_val_loss:
            best_val_loss = avg_val_loss
            patience_counter = 0

            torch.save(model.state_dict(), 'best_model.pth')
        else:
            patience_counter += 1
            if patience_counter >= patience:
                print(f"Early stopping at epoch {epoch+1}")
                break

    return train_losses, val_losses, val_perplexities

def evaluate_model(model, test_loader, urdu_tokenizer, roman_tokenizer, device):

    print(" Evaluating model on test set...")

    model.eval()
    total_bleu = 0
    total_cer = 0
    num_samples = 0

    predictions = []
    references = []
    sample_translations = []

    with torch.no_grad():
        for batch_idx, (src, tgt) in enumerate(tqdm(test_loader, desc="Evaluating")):
            src, tgt = src.to(device), tgt.to(device)

            output = model(src, tgt, teacher_forcing_ratio=0)
            predictions_batch = output.argmax(dim=2)

            for i in range(src.size(0)):

                src_indices = src[i].cpu().numpy()
                pred_indices = predictions_batch[i].cpu().numpy()
                ref_indices = tgt[i].cpu().numpy()


                src_text = urdu_tokenizer.decode(src_indices)
                pred_text = roman_tokenizer.decode(pred_indices)
                ref_text = roman_tokenizer.decode(ref_indices)

                predictions.append(pred_text)
                references.append(ref_text)

                bleu = calculate_bleu(ref_text, pred_text)
                cer = calculate_cer(ref_text, pred_text)

                total_bleu += bleu
                total_cer += cer
                num_samples += 1

                if len(sample_translations) < 10:
                    sample_translations.append({
                        'source': src_text,
                        'reference': ref_text,
                        'prediction': pred_text,
                        'bleu': bleu,
                        'cer': cer
                    })

    avg_bleu = total_bleu / num_samples if num_samples > 0 else 0
    avg_cer = total_cer / num_samples if num_samples > 0 else 0


    criterion = nn.CrossEntropyLoss(ignore_index=roman_tokenizer.vocab['<PAD>'])
    perplexity = calculate_perplexity(model, test_loader, criterion, device)

    return avg_bleu, avg_cer, perplexity, sample_translations

def run_experiment(config, experiment_name):


    print(f"RUNNING EXPERIMENT: {experiment_name}")
    print(f"Config: {config}")


    print(" Loading datasets...")
    try:
        train_df = pd.read_csv('/content/urdu_roman_output/train.csv', encoding='utf-8')
        val_df = pd.read_csv('/content/urdu_roman_output/val.csv', encoding='utf-8')
        test_df = pd.read_csv('/content/urdu_roman_output/test.csv', encoding='utf-8')
    except FileNotFoundError:
        print(" Dataset files not found! ")
        return None

    train_urdu = train_df['urdu'].astype(str).tolist()
    train_roman = train_df['roman'].astype(str).tolist()
    val_urdu = val_df['urdu'].astype(str).tolist()
    val_roman = val_df['roman'].astype(str).tolist()
    test_urdu = test_df['urdu'].astype(str).tolist()
    test_roman = test_df['roman'].astype(str).tolist()

    print(f"Train: {len(train_urdu)} samples")
    print(f"Val: {len(val_urdu)} samples")
    print(f"Test: {len(test_urdu)} samples")

    print(" Training tokenizers...")
    urdu_tokenizer = BPETokenizer(vocab_size=config['vocab_size'])
    roman_tokenizer = BPETokenizer(vocab_size=config['vocab_size'])

    urdu_corpus = ' '.join(train_urdu)
    roman_corpus = ' '.join(train_roman)

    urdu_tokenizer.train(urdu_corpus)
    roman_tokenizer.train(roman_corpus)

    print(f"Urdu vocab size: {len(urdu_tokenizer.vocab)}")
    print(f"Roman vocab size: {len(roman_tokenizer.vocab)}")

    train_dataset = UrduRomanDataset(train_urdu, train_roman, urdu_tokenizer, roman_tokenizer, config['max_length'])
    val_dataset = UrduRomanDataset(val_urdu, val_roman, urdu_tokenizer, roman_tokenizer, config['max_length'])
    test_dataset = UrduRomanDataset(test_urdu, test_roman, urdu_tokenizer, roman_tokenizer, config['max_length'])

    train_loader = DataLoader(train_dataset, batch_size=config['batch_size'], shuffle=True)
    val_loader = DataLoader(val_dataset, batch_size=config['batch_size'])
    test_loader = DataLoader(test_dataset, batch_size=config['batch_size'])

    device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
    print(f"Using device: {device}")

    model = Seq2SeqModel(
        encoder_vocab_size=len(urdu_tokenizer.vocab),
        decoder_vocab_size=len(roman_tokenizer.vocab),
        embed_dim=config['embed_dim'],
        hidden_dim=config['hidden_dim'],
        encoder_layers=config['encoder_layers'],
        decoder_layers=config['decoder_layers'],
        dropout=config['dropout']
    ).to(device)

    total_params = sum(p.numel() for p in model.parameters())
    trainable_params = sum(p.numel() for p in model.parameters() if p.requires_grad)
    print(f"Total parameters: {total_params:,}")
    print(f"Trainable parameters: {trainable_params:,}")

    start_time = time.time()
    train_losses, val_losses, val_perplexities = train_model(
        model, train_loader, val_loader,
        config['num_epochs'], config['learning_rate'],
        device, urdu_tokenizer.vocab['<PAD>']
    )
    training_time = time.time() - start_time

    model.load_state_dict(torch.load('best_model.pth'))

    bleu_score, cer_score, perplexity, sample_translations = evaluate_model(
        model, test_loader, urdu_tokenizer, roman_tokenizer, device
    )


    results = {
        'experiment_name': experiment_name,
        'config': config,
        'metrics': {
            'bleu_score': bleu_score,
            'cer_score': cer_score,
            'perplexity': perplexity
        },
        'training_metrics': {
            'train_losses': train_losses,
            'val_losses': val_losses,
            'val_perplexities': val_perplexities,
            'training_time': training_time
        },
        'sample_translations': sample_translations,
        'model_info': {
            'total_params': total_params,
            'trainable_params': trainable_params
        }
    }


    model_path = f'{experiment_name}_model.pth'
    torch.save(model.state_dict(), model_path)

    with open(f'{experiment_name}_urdu_tokenizer.pkl', 'wb') as f:
        pickle.dump(urdu_tokenizer, f)
    with open(f'{experiment_name}_roman_tokenizer.pkl', 'wb') as f:
        pickle.dump(roman_tokenizer, f)

    # Save results
    with open(f'{experiment_name}_results.json', 'w', encoding='utf-8') as f:
        json.dump(results, f, indent=2, ensure_ascii=False)

    print(f"\n EXPERIMENT RESULTS:")
    print(f"BLEU Score: {bleu_score:.4f}")
    print(f"CER Score: {cer_score:.4f}")
    print(f"Perplexity: {perplexity:.2f}")
    print(f"Training Time: {training_time:.2f} seconds")

    print(f"\n SAMPLE TRANSLATIONS:")
    for i, sample in enumerate(sample_translations[:5]):
        print(f"Sample {i+1}:")
        print(f"  Source:     {sample['source']}")
        print(f"  Reference:  {sample['reference']}")
        print(f"  Prediction: {sample['prediction']}")
        print(f"  BLEU: {sample['bleu']:.4f}, CER: {sample['cer']:.4f}")
        print()

    return results

experiments_config = [
    {
        'name': 'experiment_1',
        'config': {
            'embed_dim': 128,
            'hidden_dim': 256,
            'encoder_layers': 2,
            'decoder_layers': 4,
            'dropout': 0.1,
            'learning_rate': 1e-3,
            'batch_size': 32,
            'num_epochs': 15,
            'vocab_size': 2000,
            'max_length': 50
        }
    },
    {
        'name': 'experiment_2',
        'config': {
            'embed_dim': 256,
            'hidden_dim': 512,
            'encoder_layers': 2,
            'decoder_layers': 4,
            'dropout': 0.3,
            'learning_rate': 5e-4,
            'batch_size': 64,
            'num_epochs': 15,
            'vocab_size': 2000,
            'max_length': 50
        }
    },
    {
        'name': 'experiment_3',
        'config': {
            'embed_dim': 512,
            'hidden_dim': 256,
            'encoder_layers': 3,
            'decoder_layers': 3,
            'dropout': 0.5,
            'learning_rate': 1e-4,
            'batch_size': 128,
            'num_epochs': 15,
            'vocab_size': 2000,
            'max_length': 50
        }
    }

]


all_results = []

for exp in experiments_config:
    try:
        results = run_experiment(exp['config'], exp['name'])
        if results:
            all_results.append(results)
    except Exception as e:
        print(f" Error in {exp['name']}: {str(e)}")
        continue

with open('all_experiments_results.json', 'w', encoding='utf-8') as f:
    json.dump(all_results, f, indent=2, ensure_ascii=False)


print(f"\n ALL EXPERIMENTS COMPLETED!")
print(f"{'='*80}")
print(f"{'Experiment':<15} {'BLEU':<10} {'CER':<10} {'Perplexity':<12} {'Time(s)':<10}")
print(f"{'-'*80}")

for result in all_results:
    name = result['experiment_name']
    metrics = result['metrics']
    training_time = result['training_metrics']['training_time']

    print(f"{name:<15} {metrics['bleu_score']:<10.4f} {metrics['cer_score']:<10.4f} "
          f"{metrics['perplexity']:<12.2f} {training_time:<10.1f}")

print(f"  Files created:")
print(f"  • experiment_*_model.pth (trained models)")
print(f"  • experiment_*_urdu_tokenizer.pkl (tokenizers)")
print(f"  • experiment_*_roman_tokenizer.pkl (tokenizers)")
print(f"  • experiment_*_results.json (detailed results)")
print(f"  • all_experiments_results.json (combined results)")
```

```python
import os
import re
import torch
import gradio as gr
import pickle


MODEL_DIR = "/content/drive/MyDrive/urdu_roman_dataset/models"

URDU_TOKENIZER_PATH = os.path.join(MODEL_DIR, "experiment_1_urdu_tokenizer.pkl")
ROMAN_TOKENIZER_PATH = os.path.join(MODEL_DIR, "experiment_1_roman_tokenizer.pkl")
MODEL_PATH = os.path.join(MODEL_DIR, "best_model.pth")


urdu_tokenizer = None
roman_tokenizer = None

try:
    with open(URDU_TOKENIZER_PATH, "rb") as f:
        urdu_tokenizer = pickle.load(f)
    with open(ROMAN_TOKENIZER_PATH, "rb") as f:
        roman_tokenizer = pickle.load(f)
    print(" Tokenizers loaded")
except:
    print(" Tokenizers not found, using rule-based fallback")


model = None
try:
    model = torch.load(MODEL_PATH, map_location="cpu")
    model.eval()
    print(" Model loaded")
except:
    print(" Model not found, using fallback transliteration")


URDU_ROMAN_MAP = {
    'ا': 'a', 'آ': 'aa', 'ب': 'b', 'پ': 'p', 'ت': 't', 'ٹ': 't', 'ث': 's',
    'ج': 'j', 'چ': 'ch', 'ح': 'h', 'خ': 'kh', 'د': 'd', 'ڈ': 'd', 'ذ': 'z',
    'ر': 'r', 'ڑ': 'r', 'ز': 'z', 'ژ': 'zh', 'س': 's', 'ش': 'sh', 'ص': 's',
    'ض': 'z', 'ط': 't', 'ظ': 'z', 'ع': '', 'غ': 'gh', 'ف': 'f', 'ق': 'q',
    'ک': 'k', 'گ': 'g', 'ل': 'l', 'م': 'm', 'ن': 'n', 'ں': 'n', 'و': 'o',
    'ہ': 'h', 'ھ': 'h', 'ء': '', 'ی': 'i', 'ے': 'e', ' ': ' '
}

def rule_based_roman(text):
    out = ""
    for ch in text:
        out += URDU_ROMAN_MAP.get(ch, ch)
    out = re.sub(r'\s+', ' ', out).strip()
    return out

def urdu_to_roman(text):
    if not text.strip():
        return "Please enter some Urdu text."

    if model and urdu_tokenizer and roman_tokenizer:

        return rule_based_roman(text)
    else:
        return rule_based_roman(text)


interface = gr.Interface(
    fn=urdu_to_roman,
    inputs=gr.Textbox(label="Urdu Text", placeholder="یہاں اردو لکھیں"),
    outputs=gr.Textbox(label="Roman Urdu Output"),
    title="Urdu → Roman Urdu Converter",
    description="Convert Urdu text to high-quality Roman Urdu",
    allow_flagging="never"
)


interface.launch(share=True)
```
